```python
import pandas as pd
df = pd.read_csv("/content/sample_data/voice (1).csv")

print("Ukuran dataset:", df.shape)
print(df.head())

print("\nJumlah data setiap kelas:")
print(df["label"].value_counts())
```

```python
X = df.drop("label", axis=1)

y = df["label"]

print("Jumlah fitur:", X.shape[1])
print("Nama fitur:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
X_train_70, X_test_70, y_train_70, y_test_70 = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Data training:", len(X_train_70))
print("Data testing :", len(X_test_70))
```

```python
svm_linear_70 = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="linear"))
])
svm_linear_70.fit(X_train_70, y_train_70)

y_pred_linear_70 = svm_linear_70.predict(X_test_70)

accuracy_linear_70 = accuracy_score(
    y_test_70,
    y_pred_linear_70
)

print("Akurasi SVM Linear 70:30 =", accuracy_linear_70)
```

```python
svm_poly_70 = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="poly"))
])

svm_poly_70.fit(X_train_70, y_train_70)

y_pred_poly_70 = svm_poly_70.predict(X_test_70)

accuracy_poly_70 = accuracy_score(
    y_test_70,
    y_pred_poly_70
)

print("Akurasi SVM Polynomial 70:30 =", accuracy_poly_70)
```

```python
svm_rbf_70 = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf"))
])

svm_rbf_70.fit(X_train_70, y_train_70)

y_pred_rbf_70 = svm_rbf_70.predict(X_test_70)

accuracy_rbf_70 = accuracy_score(
    y_test_70,
    y_pred_rbf_70
)

print("Akurasi SVM RBF 70:30 =", accuracy_rbf_70)
```
